# Plot

```python
df = pd.read_csv('degradation result for plot.csv')
df = df[['Description Length (% Sentences)', 'Original','Random Order', 'Reordered Sentences STDEV', 'Reversed Order', 'Sides to Middle']]
# change 'Sides to Middle' to 'Sides-to-Middle'
df.rename(columns={'Sides to Middle': 'Sides-to-Middle'}, inplace=True)
df
```

```python
# plot the results of the degradation test
plt.figure(figsize=(9, 6))
plt.errorbar(df['Description Length (% Sentences)'], df['Original'], label='Original', marker='o', linewidth=3, markersize = 8)
plt.errorbar(df['Description Length (% Sentences)'], df['Random Order'], label='Random Order', marker='o', yerr=df['Reordered Sentences STDEV'], capsize=5, linewidth=3, markersize = 8)
plt.errorbar(df['Description Length (% Sentences)'], df['Reversed Order'], label='Reversed Order', marker='o', linewidth=3, markersize = 8)
plt.errorbar(df['Description Length (% Sentences)'], df['Sides-to-Middle'], label='Sides-to-Middle', marker='o', linewidth=3, markersize = 8)

plt.xlabel('Description Length (% Sentence Inclusion)', fontsize=20)
plt.ylabel('MAE (eV)', fontsize=20)
plt.title('LLM-Prop Prompt Length and Order Effect', fontsize=22)
# legend with the order of the plots
plt.legend(fontsize=16, loc = 'upper center')
# x-axis ticks
plt.xticks(np.arange(10, 110, 10), fontsize=20)
# y-axis ticks
plt.yticks(fontsize=20)
```
